### Reflection

This book is only getting more and more interesting! Chapter 15 opens up Part 3 of the book and discusses a hotly debated topic--sentience in AI. However, this idea of sentience is difficult to attribute to humans consistently, as it can be seen as a temporal state. In addition, it is difficult to label AI and LLM systems as sentient because all they do is regurgitate humanlike language based on training data. In fact, it is human inclinations and subjectivity that attribute sentience to the LLMs. Summerfield then goes on to discuss this concept of "intentionality" and whether LLMs are capable of it. Intentionality is defined as a mental state *about* something, being to know, to desire, to believe, or to understand something. LLMs tend to express their output in such ways; however, this is not an indicator of intentionality in LLMs. 

In chapter 17, Summerfield explores the logical "mind" of LLMs, their ability to solve logical puzzles, correct their mistakes, and give accurate knowledge. However, these are often due to the probabilistic nature of LLMs, which can "reason" to some degree; however, this does not mean that the LLM mind works in the same way as humans do. Summerfield would compare LLM processing to "parrot" talk where its words are ultimately copied from humans--with this in mind, remembering that LLMs are not human-like in itself but only in the way it imitates human speaking patterns. Summerfield then raises rising concerns for the humanization of LLMs, claiming it as harmful to humanity for even simply making this comparison. Another critique of the perpetuation of human structural inequities in technology, however, as I would also agree, other forms of structural inequality need to be more urgently addressed first.  

Ultimately, part 3 of *These Strange New Minds* explores the complexities of LLM behavior and output. With large amount of data, training, and fine-tuning, LLMs are capable of performing "intelligent" tasks, whether or not you believe they are capable of this intelligence or not. Summerfield asserts his position that LLMs are in fact capable of this act of thought, which in part, contributes to their status as intelligent systems. And after reading these past few chapters, I would say that I've come to agree with his stance as it would be difficult to ignore the intellectual processing of information LLMs are capable of and not say they are actively reasoning and thinking.

Summerfield emphasizes the projection of human thought and emotions onto these technological systems. This is such an intriguing idea, and I think after reading this book, I've come to solidify my understanding of how these LLM systems work and how human-like but inhuman they are. I agree with Summerfield's perspective that LLMs are being used as scapegoats for other existing societal concerns, such as structural inequity due to rising fear of technology, and how unfounded or misdirected these concerns often are. I do wish that Summerfield had elaborated more on this idea of how some groups of people may be overreliant or overproject human-like qualities onto LLM systems. I've seen stories online about how some people have fallen in love with LLM systems or digital avatars, and I'd be very interested to see how Summerfield views these experiences and rationalizes them.
