### Reflection

Chapter 7 begins by dissecting this idea of language acquisition and how natural language processing (NLP) researchers are able to encode human language. Summerfield then introduces the idea of apes learning sign language in chapter 8. Nim was a chimpanzee who was taught how to sign ASL. However, this project would eventually go horribly wrong as Nim grew to be aggressive if his signed demands were not met, and he grew a temper. Despite this, Nim wasn't "learning language" the same as we do, but rather identified associations between signs and his desires--making language communication something very primal to him. Nim could not reproduce signing with the same syntactical understanding as humans do.

However, although I understand the author's point of view on this topic, I feel like it can still be stated that Nim, along with the other ASL proficient animal companions brought up in this chapter, were still understanding the language--and if anything, perhaps it developed its own dialect of ASL just as language diverges to adapt to a specific culture. The quasi-random order of Nim's ASL may be a reflection of this evolution. I wish this idea was explored a little deeper because I believe that syntactical understanding may be presenting differently as it is in humans--but that does not negate the chimpanzee's understanding of it.

The next two chapter deep dive into ELIZA--a chatbot used in psychotherapy that mimicked natural dialogue using scripts and matching plausible inputs to outputs. It is certainly a fascinating NLP that so clearly does algorithmic matching that it doesn't appear to "understand" as we traditionally see fit, but the illusion of it stands so very strong. This was particularly interesting to learn about because the idea of speaking to an LLM endlessly for therapeutic uses has fascinating medical implications. I'm looking forward to reading more about it in the future and getting through the next few chapters.
