### Reflection

Chapters 11 to 14 go into a deep dive into how these statistical models are able to predict, remember, generate, and transform information to then create grammatically coherent and precise language. Chapter 11 shares how these models are able to create semantic maps that organize concepts associated with words, and then organzing those concepts in relation to each other, in attempts to make text generate coherent, meaningful, and relevant to the context. As these models were initially limited and could not generate relevant text from a longer string of information, recurrent neural networks (RNNs) rose. RNNs have short-term memory that compresses input into numeric representations, which is then used to make more accurate predictions of the next word--and therefore, create coherent text. The seq2seq models that derived from the RNNs was then further refined through the development of "the transformer" in 2017. The transformer allowed these models to ignore or emphasize specific input to refine its prediction and generation--giving its text a "human touch." The relevation of the transformation took these models to the level of "human level" textual generation that was unable to be achieved previously despite astronomical amounts of data being poured into these models. 


The unraveling evidence and narrative of how LLMs have come to be as powerful as they are today in predicting and generating syntactical word sequencing was actually incredibly fascinating. These past few chapters made me the most absorbed I've ever been in this book, and it really felt like it was starting to pick up. Particularly, I felt that I was beginning to get used to the author's narrative style with how he introduces pop cultural and familiar references, and then connects to his overarching narrative about the development of these LLM and generative models. 


In addition, in these past few chapters, more of the author's position on LLMs was highlighted. The author appears to be extremely fascinated by AI and its capabilities, and appears to be on the more optimistic side of the generational capabilities of AI. As a point of opposition, the author introduces Noam Chomsky, a prestigious linguist, who is a strong critic of the statistical approach to language, arguing that the language is highly complex and innate to the human mind. While Summerfield appears to agree with Chomsky in some aspects, such as the great difficulty of producing grammatical sentences with statistical models, he counters that emerging technology such as the transformer now makes what was thought to be impossibly by Chomsky, possible.

I would definitely agree with Summerfield in this regard. In the way that Summerfield has painted Chomsky in this book, he comes off as a conservative linguist who clings to the past, claiming that AI and generational statistical models will never be able to mimic human speech accurately. However, this is a stubborn clinging to old ideologies. I believe that, especially with the rapid progression of technology development, LLMs are able to create grammatically coherent sentences and speech, and in fact, it's already doing it and sometimes even better than I can at times.
