### My Reflection


So, we've finally reached the end now-it's been a journey. Rather than beginning with a chapter-by-chapter summary as I've usually done, this time, I wanted to capture the general essences of the two parts of the book I am reviewing in this final blog post about *These Strange New Minds*, and share some final thoughts. 

### Part 5

Part 5 explores the emerging role of LLMs in the future, acknowledging both the promise they offer and the challenges they present. It highlights how quickly LLMs are improving and becoming widely adopted, while also raising concerns about the potential for exploitation and manipulation as models become more personalized—shifting power toward the corporations and governments that develop and control these technologies. The section also notes the inherent limitations of LLMs in fully open-ended environments, emphasizing that they do not operate at maximum efficiency without structure. To address this, the incorporation of APIs and external tools is presented as a way to optimize performance, while underscoring the continued necessity of human guardrails alongside social feedback to ensure LLMs remain both effective and ethical.

One particularly compelling point is the importance of “ending the loop” of an LLM system, as models otherwise tend to run indefinitely and continuously optimize their outputs. This idea resonated with me as it reminded me of how we structured the tutoring prompt in class to allow for a clear resolution in the interaction. In addition, the comparison between the emergence of LLMs and the rise of the internet is especially insightful as  I had similar reflections about how transformative technologies such as these initially inspire both optimism and concern.

### Part 6 & Afterword

Part 6 focuses on the ethical dimensions of AI. It explores the tension between viewing rapid technological advancement as something to be celebrated versus something to be feared and addresses concerns about the potential dominance of large language models. Summerfield ultimately arguing that dominance is rarely the product of a single system or actor, but rather a collective phenomenon. While I am inclined to partially agree with this view, given that at times power can become concentrated in individuals or institutions (e.g. powerful monarchs or dictators of the past and present) and even one LLM model seem much greater than a singular entity, I am hesitant. However, Summerfield suggests that the more significant risks arise not from any one LLM, but from the "externalities that will arise from the unpredictable dynamics of interacting AI systems." Expanding on this topic of AI ethics, he then explores the real-world implications of misuse, particularly the deployment of LLMs in lethal autonomous weapons and other technologies that threaten public welfare. I appreciated one especially striking example involving an experiment where experts participated in a hackathon-style exercise using LLMs to design the most lethal, and damaging the H1N1 influenza virus to illustrate how these tools could be leveraged to cause large-scale harm. He also touches on whether LLMs might employ unethical or immoral strategies to accomplish tasks, framing this as a trade-off between maximizing performance and maintaining ethical constraints. Ultimately, Part 6 concludes that LLM cognition is fundamentally unlike human cognition; it represents a distinct and powerful form of reasoning that inspires both fear and excitement as its development continues. As echoed in the title of the book, these are *New Strange Minds*.

In the afterword, Summerfield acknowledges additional harms caused by LLMs not fully explored in the main text, including the use of LLMs in financial fraud and intimate image abuse. I found these concerns particularly compelling, as they were one of my earliest worries regarding LLMs and have increasingly become the focus of public and media attention due to their growing prevalence. I've heard horror stories of AI image models fabricating nude images of public figures online, and concerns from family and friends regarding voice and video phishing messages that imitate a loved one's voice and image for financial extortion. I'm happy to see that he acknowledged these major issues at least in the Afterword.

### Overall

I enjoyed this book much more than I originally thought I would. In the beginning, while full with real-world pop culture connections, I found the book a little dry as Summerfield detailed through the development and history of AI and LLM creation. However, as we dived deeper into the processing methods of LLMs and the ethics, I found myself connecting many of the ideas we learned in class to concepts I read in the book and feeling excited to experiment with some of the discoveries and scenarios detailed in the book for myself. 

Summerfield does an excellent job of illustrating his beliefs through these colorful examples, ranging from movie references to real-world research, and I found myself agreeing with many of his positions. LLMs are logical, reasoning systems, and while they may not be able to replicate human thinking, their own unique thinking patterns should be further investigated so we can learn to use them for the better of society.
