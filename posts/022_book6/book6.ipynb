### Reflection

After wrapping up part 3 of the book with much room for thought, part 4 opens with the question, "What should a language model say?" The "language game," as Summerfield calls it, is one of the true challenges of LLMs--learning to understand nonverbal and social cues to understanding subliminal meaning in language.  Chapter 23 introduces the complex ethical concers regarding the development of LLMs--the main goal being to prevent the generation of harmful, hurtful, misleading, or nefarious content out into the real world. Due to these concerns, companies like OpenAI heavily fine-tuned and filtered the LLM outputs to ensure they adhere to their ethical guidelines. Fine-tuning refers to the process of retraining an LLM model with human feedback. This process leads to improved validity of LLM generation and decreased harmful output. Despite these benefits, however, there are some additional considerations to be taken. For one, there is a concern that LLMs may become less efficient at a given task at hand after fine-tuning. Even more significant, there is significant bias that comes from the innate beliefs of the group of people responsible for fine-tuning the model. For example, while openAI capitalizes on being non-opinionated and neutral, evidence suggests that openAi tends to have a more liberal political stance which is likely due to the political views and fine-tuning of its human reviewers.

Summerfield also explores a common concern shared in popular media--hallucination, referred to as confabulation in the book. Fine-tuning is very effective at reducing the amount of confabulated information; however, with this more limited response, LLM output may come off as more generic than not. This emphasizes the importance of detailed, structured prompting to avoid these defaulted generic answers. It also raises another similar concern regarding LLM use--misinformation. LLM is particularly persuasive and could be used for more positive or negative purposes. Whether used to educate the general public about public health strategies, or used to cause moral harm, more consideration should be given regarding the power of LLM and its ability to sway public thought, opinion, and belief.

I think these were my favorite chapters of the book by far. I really love exploring and learning more about the ethical concerns of LLMs, and these chapters focused on exactly that. Summerfield expresses concern regarding the accessibility of LLMs to the public and the possibility of exploitation from these models. Rather than framing these models as inherently dangerous or beneficial, Summerfield encourages a more careful, complex engagement of this discourse by acknowledging its power and remaining critical of the assumptions and biases embedded in them. The questions he raise are urgent, and are even something I'd like to explore in my last few blog posts.
