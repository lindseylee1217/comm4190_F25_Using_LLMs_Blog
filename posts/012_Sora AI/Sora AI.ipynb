In today's digital age, the line between reality and fabrication is becoming increasingly blurred, thanks to the rapid advancements AI-generated videos. These videos are widespread, encompassing a wide range of content—from soothing ASMR clips and entertaining animal antics to more sinister applications that include spreading misinformation and creating lifelike videos of political figures. In this blog, I wanted to survey my friends and see whether they would believe my AI video was real.

### Survey Results & Reflection

Overall, all 7 of my friends were able to guess that the video was AI-generated. Many of them pointed out that the "video looked fake" because of the textures in the video. Despite being able to accurately identify that it was an AI video, many of them had also exclaimed on the realism of the video and how great the quality was. 

If my friends had been any less keen on technology, I think it would have been easy to fool someone into thinking that this was a real video with real animals. Even more, if technology continues to improve at this rapid rate, it's possible that humans are no longer able to distinguish between AI and real videos anymore.

One thing I also wonder that I hope to investigate more in the future is what source and training materials LLMs use to generate these videos. I'd be curious learn more about how this source material influences the type of product generated.
