<h3 style="background-color: #E0F4FF;"> Inspiration </h3>

My professor has recently been incorporating AI into our into our in-class participation to increase our literacy for the intersections between AI and healthcare. She has been accomplishing this by giving us prompting exercises and reflecting on the AI output, noting its strengths and weaknesses.

As study material for my exam, I decided to write a case study to practice my knowledge on the material. Referencing my study notes, I chose a concept to base my case study on and came up with the following:

> A 45-year-old female patient was admitted to the ED after suffering an automobile accident. She had an exploratory laparotomy and is now post-op day 1. Patient reports increasing abdominal pain of 8/10, stiff abdomen, and decreased urine output.
> - What assessments should you prioritize at this time?
> - What complication do you suspect may be occuring?
> - Given the complication, what actions should be prioritized?

This question is based on material related to abdominal trauma and considers nursing assessment, interventions, and prioritizations within the question.

My response to this case study would be:

> **Assessments**: Focused GI assessment, obtain vital signs, check labs, measure intra-abdominal pressure, monitor intake and output
> 
> **Complication**: Abdominal Compartment Syndrome
> 
> **Actions**: Immediately move the patient to the OR for immediate surgical decompression. 

<h3 style="background-color: #E0F4FF;"> LLM Output </h3>

I asked gpt-4o the prompt and added the following modification to the beginning:

> Imagine you are a nursing student. Please answer the following case study.

#### **Assessment**

It gave me the following output:

> Vital signs, abdominal assessment, pain assessment, fluid status, urine output, general appearance and behaviors

This is quite a comprehensive list of focused assessments to complete on this patient. However, I would add that it would be important to monitor the patient's labs and intra-abdominal pressure to rule out the potential of other complications and monitor the patient's condition. 

#### **Complication**

It gave me the following output:

> The symptoms described—severe abdominal pain, a stiff abdomen, and decreased urine output—suggest postoperative ileus or possibly more critically, abdominal compartment syndrome or internal bleeding. The stiff abdomen and pain are particularly concerning and might indicate peritonitis or an intra-abdominal complication.

Gpt-4o was able to successfully narrow down the given symptoms and signs into a list of reasonable complications. The complication I was writing about was abdominal compartment syndome, a condition in which increase intra-abdominal pressure due to blunt or penetrating trauma, or fluid resusitation causes impaired circulation and organ dysfunction. The signs and symptoms the patient presents with--increasing pain, stiff abdomen, and decreased urine output--are characteristic of abdominal compartment syndrome. However, I can see how gpt-4o, without knowing the scope of the material of my class and emphasis on certain pathophysiologies, could have reasonably determined that these signs and symptoms were reflective of other conditions. 

#### **Actions**

It gave me the following output:

> 1. Notify the Physician/Surgeon
> 2. Prepare for Diagnostic Tests
> 3. Monitor and Maintain IV Fluids
> 4. Administer Pain Relief as Prescribed
> 5. Reassess Frequently
> 6. Educate and Reassure the Patient

I was a little unsatisfied with this response as I felt that gpt-4o didn't address the question in its entirety. It seems that gpt-4o gave a list of actions to narrow down the complication even further, rather than coming up with a list of individualized actions tailored to each complication it identifies. Given that the complication is abdominal compartment syndrome, which gpt-4o was able to partly identify, the immediate action would be to prepare the patient for surigcal operation because abdominal compartment syndrome is a medical emergency. While gpt-4o was able to identify the significance of abdominal compartment syndrome as cited above in its analysis of the complication the patient is experiencing, it was unable to translate this priority in its actions. Perhaps with more guidance gpt-4o would have been able to identify that transitioning the patient to the operating room would take utmost priority. 

This may have been a prompting error as well, and the instructions could have been more clear, however, I believe that to the human eye, the question is clear enough.

---

<h3 style="background-color: #E0F4FF;"> Reflection </h3>

My main takeaway from this exercise is that while gpt-4o is certainly good at generating ideas and narrowing down the solutions of questions, it won't be replacing human critical thinking in the medical field. Gpt-4o was able to synthesize information very efficiently and clearly and was able to get most things related to my study material--emphasis on *most* things. However, I don't think my own knowledge is perfect either, so I believe that gpt-4o's output in this example showcases the potential of LLMs as tools in healthcare and facilitators in guiding more efficient nursing critical thinking.
