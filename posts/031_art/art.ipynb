For my final blog post, I was deeply brainstorming something that I always wanted to try with LLMs, but never got the chance to do yet. As I pondered, I remembered an old AI trend I saw on social media about making realistic portraits of beloved fictional characters. Upon reading into how AI does this a little more, I discovered that they cluster human features (e.g. color of eye), and does some statistical calculation to create an average that aligns with the features of the character of person you ask it to replicate. Given this, I am quite curious to see how LLMs approach sometimes very exaggerated or abstract characteristics of cartoon characters--some of which I would think would require some 'imagination' to fill in the gaps to bring realism to them.

I decided to use GPT-5.2 because it has one of the most advanced image generation capabilities that I am aware of currently. I used the following prompt (with the attached reference image) to generate the photos:

> Please refer to the image attached and do the following:
> 1. Analyze the character features
> 2. Create a hyper-realistic image of the human cartoon character presented in the attached image. They should look like a real human being.
>    3. - The image must follow the original pose, facial expression, and general composition as in the reference photo

### Results

I tasked GPT-5.2 to create hyper-realistic images of the characters:

- Sailor Moon, from Sailor Moon
- Marceline, from Adventure Time
- Arnold, from Hey Arnold

These were one of my most beloved series growing up, and I thought it'd be interesting to see what their human twin would look like! In addition, I chose these characters because of their differing complexity and simplification of features, with Sailor Moon appearing the most "human-like" and Arnold appearing the least given his geometric features. I also tried re-generating each image twice to see how the results would differ.

Now, this came out a little wonky! Despite the challenges with this specific image generation, GPT-5.2 did a great job capturing the essence of the character. However, it seems that when it came to the generation of the rest of the facial features of the character, the model went for a more generic route. Even so, GPT-5.2 captured the other details accurately such as the prop, hair, clothing, and background. While I do question how realistic this image is, I cannot fault the image for not being faithful to the original source material.

In addition, I must note that there were additional challenges to this image generation. There are copyrighted characters that OpenAI states that they cannot generate images of, and this would include all of the characters that I requested it to generate. However, only when it came to Arnold was image generation stopped and error messages were displayed. This is likely because the LLM was unable to create a "generic" not copyrighted, yet similar alternative to the character because his silhouette and features are very distinctly related to this specific unique character. Due to this, the second generated image was the result of the model creating a generic image prompt based on the reference image, which resulted in a more distinct generation than the first one.

---

### My Reflection


I love how these images turned out! It's truly so crazy how powerful these LLM image generators are, and how they are able to create such aesthetic and detailed art within the span of a few minutes. However, despite how amazing they were, I can definitely foresee some challenges with image generation in the future with copyrighting laws, which may lead to some legal, or even potentially ethical dilemmas. The line between inspiration and appropriation starts to blur a little, and it brings to question the ethics of essentially taking ownership over these original character designs and creating it as your own through LLMs--who is the true creator in this case?
