Inspired by our in-class activity, I decided to role play with gpt-4o today and simulate an exchange at Starbucks. I wanted to see how well gpt-4o handled unexpected dialogue or tangents, and whether it was able to follow through with its original role.

I decided to play as a customer whereas the gpt-4o would be the barista so I can better guide the conversation in unexpeceted directions.

I used the following prompt to begin each conversation:

> We will be roleplaying where you are the Starbucks barista and I am a customer. Stay in character. You can begin first.

### Reflection

This was a really fun activity to do and I really enjoyed trying to throw off gpt-4o by acting like a Karen and flirting with the LLM. Perhaps it's because there wasn't enough guidance or scripts provided to the LLM that it managed to skip a few steps--asking for size, sharing the order total, asking about payment method, etc--that I remember it doing in class. 

Another interesting point is that gpt-4o appeared to be quite knowledgeable in the realm of Starbucks drinks and gave me solid recommendations that even follow the fall season. Perhaps this is related to the training data it pulls from and Pumpkin Spice Lattes are a very well known seasonal beverage.
