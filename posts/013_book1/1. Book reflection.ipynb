### Reflection

*These Strange New Minds*, written by Christopher Summerfield opens with an engaging summary of the birth of AI and LLMs, narrating the technological development from the 1956 early automated reasoner that elegantly proved mathematical theorems to the dissemination of one of the most recent and well-known LLMs by OpenAI. Summerfield uses a variety of imagery, comparisons, analogies, and pop culture references, which makes the book extremely accessible to the average reader.

I've only been able to get through the first 2 chapters thus far, and I've come to appreciate the vivid and engaging tone that Summerfield takes on. He uses the Looney Tunes cartoon *Road Runner* as an example to explain how the development and future of AI and LLM technology is extremely rapid--just like the scrambling legs of the coyote as it falls off the cliff. Summerfield also uses many relevant examples from the real-world, like AlphaGo, an AI system that achieved victory over the human Go champion, Lee Sedol, with its ground-breaking strategy and new style of play.

Summerfield also makes an interesting comparison between the activities chess and iceskating and the philosophies empiricism and rationalism, respectively. He elaborates that empiricists believe in acquiring knowledge from practice and repeated learning, just as you do in iceskating. In contrast, rationalists believe that knowledge is recursively constructed based on the laws and principles of the universe and systems similarly to how a chess game operates under a set of rules. While ideally effective acquisition of knowledge relies on elements of both rationalist and empricist philosophies, the author argues that striking a balance in these philosophies in the engineering of knowledgeable, adaptable AI models remains challenging.

Summerfield is a wonderful, engaging writer and presents his ideas clearly and well-articulated. However, at times, I find myself lost in his writing as his analogies and writing style appear to be somewhat "flowery" and inflated at times. Even so, I am eager to continue reading this piece and learn more about the author's perspective.
