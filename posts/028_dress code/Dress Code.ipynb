Dress codes are an interesting social construct created in society that varies much depending on sociocultural context, beliefs, and values. Given this, I was curious whether LLMs had the ability, or were at least trained to be able to discern the nuances of different dress codes or even simply identify different articles of clothing in an image. So, in this blog post I plan to do exactly this.

### Method

I wanted to employ the "thinking out loud" strategy for LLM generation, so as the first step, I wanted GPT-5.2 to describe different types of dress codes. I gave it the following prompt to start this conversation:

> Give me a 1-2 sentence description of the following types of dress codes:
> - Casual
> - Business Casual
> - Business Formal
> - Black Tie
>
> Please give specific information about unique articles of clothing, style essences, colors, fabric types, or structure that is unique to each.
>
> Please remember this information going forward.

Then, I grabbed three different images from Pinterest, each of different dress codes that would capture the essence of these descriptions. 

Lastly, I provided the image to GPT-5.2 and gave it the following prompt:

> Given the information you retrieved above about different dress codes, please do the following:
>
> 1. Describe the outfit in the image (Identify any articles of clothing you can discern).
> 2. Given the description, match the outfit to one of the following dress codes: casual, business casual, business formal, and black tie.

### Dress Code Descriptions

These are the descriptions of the dress code as generated by GPT-5.2:

> **Casual**: Relaxed, comfort-first pieces like jeans/chinos, tees, casual button-downs, sweaters, and sneakers; fabrics are often denim, cotton, fleece, or knits, with flexible silhouettes and any colors/prints (including bold or streetwear).
>
> **Business Casual**: Polished but not suit-level—think chinos or tailored trousers with a collared shirt/blouse, knit polo, or sweater, plus loafers/clean leather sneakers; neutral/soft palettes (navy, gray, tan, white) and structured-but-easy fabrics like twill, oxford cloth, merino knits, or ponte.
>
> **Business Formal**: Classic office suiting: a matching suit (or structured dress/skirt suit) with a crisp dress shirt/blouse, leather dress shoes, and minimal accessories; darker, conservative colors (navy/charcoal/black) and sharper tailoring in wool/worsted fabrics with clean lines.
>
> **Black Tie**: Eveningwear with high formality—traditionally a black (or midnight navy) tuxedo with satin/grosgrain lapels, white formal shirt, black bow tie, cummerbund or waistcoat, and patent leather shoes (or a floor-length evening gown); luxe fabrics and finishes like wool barathea/velvet and satin details are the signature.

### Image #1: Business Casual



**LLM Output:**

> 1. Outfit description (articles of clothing):
> - Dark brown oversized blazer/sport coat
> - Black crewneck top (looks like a knit or tee)
> - High-waisted taupe/brown wide-leg tailored trousers with a black belt
> - Black loafers/flat shoes
> - Black shoulder bag (and what looks like dark gloves or long sleeves peeking out)
> 
> 2. Dress code match: Business Casual — it’s polished and tailored (blazer + trousers + loafers) but not a full matching suit or formal shirt/tie, so it reads elevated office-appropriate rather than business formal or black tie.

It is notable to say that even after multiple re-generations, GPT-5.2 gave the same output, and it was in fact, correct to say that this was business casual!

### Image #2: Black Tie



**LLM Output:**

> 1. Outfit description
> - Dress: A floor-length black evening gown with an asymmetrical/one-shoulder neckline (one shoulder exposed), a fitted bodice/waist, and a sleek, draped skirt with a small train or cape-like panel at the back/side.
> - Accessories: Minimal jewelry (small earrings/bracelet); no visible bag.
> 
> 2. Dress code match
> - Black Tie. The long formal gown, structured silhouette, and minimalist elegance fit black-tie standards far more than business formal or any casual category.

Correct again! I appreciated how ChatGPT-5.2 was able to analyze the outfit and fit it into a certain dress code.

### Image #3: Casual



**LLM Output:**

> 1. Outfit description
> - Top: A fitted white turtleneck (looks like a ribbed knit) layered with a plaid button-up/overshirt peeking through.
> - Outerwear: An oversized brown jacket (workwear/bomber vibe) with a relaxed, slouchy silhouette.
> - Bottom: A brown plaid pleated mini skirt, styled with a belt at the waist.
> - Shoes: Black knee-high boots with a chunky, structured look.
> - Accessories: Layered necklaces (including a cross pendant) and a black backpack/bag.
> 
> 2. Dress code match
> - Casual. The mini skirt + oversized jacket + knee-high boots reads as styled streetwear/casual rather than office-appropriate tailoring (business casual/formal) or eveningwear (black tie).

I thought this would have been a little more difficult because the style can be seen as more preppy, and potentially more business casual, but ChatGPT-5.2 was unable to be fooled at all and correctly identified it as a casual outfit.

### My Reflection


After learning so much about how LLMs are essentially "deprive of sensory data" and experiences as mentioned in the book, *These Strange New Minds*, it was very interesting to see how ChatGPT-5.2 exceled at identifying dress codes based off an image. This gives me much promise--for the better or worse--that LLMs are not limited by their inability to experience the real-world as humans are, as they are still able to obtain so much information through images and other secondary sources of information. This was a fun experiment!
