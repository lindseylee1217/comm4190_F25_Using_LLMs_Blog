### Reflection

The book has begun to pick up! Chapter 3 started with a bang, explaining the foundational concepts of booleans and how complex reasoning can be broken down into "and," "or" and "not." Summerfield then goes on to elaborate on several other methods of "breaking down reasoning" and creating logical systems to organize data that may be applied to models to imitate human thinking. He narrates that there was a shift in thinking on how to build human intelligence into machines, moving away from making building machines "that reason about the world as if it were a huge coherent, logical system" which reflects rationalist philosophy. He makes this point under the pretext of a larger theme that the world in itself is an "arbitrary and incoherent" place. There are always exceptions made in human logical derivations of the universal laws, and the world remains an unpredictable place. 

Then chapter 4 explores the birth of the neural network and deep learning, a process that involved training these networks of multiple layers to follow a chain of operations that would allow neural networks to perform in ways it was never able to before. Summerfield further explores this concept in chapter 5 where he explains the significance of a model's ability to generalize--an important goal of deep learning. That is, generalizing understanding of existing knowledge, generalizing to generate new predictions. This is quite an astonishing distinction of LLMs as they are able to do this very skill very well. For example, chapter 6 shares how an immensely large training database and gradient descent made translations between English to Latin, and English to Thai fairly accurate. And given the LLMs ability to generalize, the model was also able to eventually translate between Latin and Thai even without being specifically trained on it. This is such an interesting concept that warrents further exploration.

In addition, one other thing that interests me is how the author, despite the rather informative purpose of the text, consistently engages the audience with humor and small jokes, attempting to balance the technical jargon with humorous anecdotes and comparisons to draw the audience back in. I can infer that the author is an intellectual who appreciates teaching others who may not be as familiar with the subject of AI and LLMs. He makes his writing approachable so it is easy to digest this new information. 

As of now, I'm excited to dive into part 2 of the book and see what new commentary and history that author can share!
