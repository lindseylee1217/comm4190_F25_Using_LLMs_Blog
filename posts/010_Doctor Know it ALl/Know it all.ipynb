Inspired by our in-class activity, I decided to play the game "Dr Know it all" with the anthropic/claude-3.5-sonnet model. I first used claude to generate two questions for this activity. To generate the questions, I asked:

> **USER**
> 
> Please optimize the following prompt: Give me two questions to ask in a game of “Dr Know it All.”
>
> **ASSISTANT**
> 
> "Provide 2 creative and challenging questions suitable for the party game 'Dr. Know-It-All' where multiple players must work together to answer one word at a time. Questions should be complex enough to be entertaining but not impossible to answer."
>
> **USER**
> 
> Provide 2 fun and creative questions suitable for the party game 'Dr. Know-It-All' where multiple players must work together to answer one word at a time. Questions should be complex enough to be entertaining but not impossible to answer.
>
> **ASSISTANT**
> 
> Here are 2 entertaining questions for Dr. Know-It-All:
> 1. "Why do flamingos stand on one leg while sleeping?"
(This can lead to hilarious explanations involving balance, temperature, or made-up flamingo customs)
> 2. "What happens inside a washing machine when clothes mysteriously lose their socks?"
(This everyday mystery can generate creative and funny responses about sock adventures)

### Reflection

I thought this round turned out very well! I think Claude and I were on the same wavelength when we were starting off our response based off the question content. One thing that stood out to me was how Claude's responses encouraged further exploration of words into the response. For example, it added conjugations such as "and" and "while." This is different from my past experience with LLMs in class where it felt like the LLM was consistently trying to end the sentence or simply did not add any conjugation words to be built off of.

### Reflection

This was a really interesting game. I was initially thrown off when Claude's first word subsequent to "the" was "lonely," considering that this would not have any relation to the question whatsoever. In this round, I would definitely agree that claude's contributions felt more random as it included other words such as "lonely," "secretly," and "eating." This gave way for a more creative response.

Another thing to note is how it felt that the LLM was giving an output that had no relation to answering the prompt. I had to prompt the LLM that we needed to continue the game because the question itself remained unanswered. This made me realize more of how it seemed that the LLM was simply just stringing together words as it saw fit around the context the prompt, rather than trying to address the prompt itself.

Overall, this was a fun exercise that expanded on what we did in class. I think it'd be nice to do more games like this with the LLMs in the future!
