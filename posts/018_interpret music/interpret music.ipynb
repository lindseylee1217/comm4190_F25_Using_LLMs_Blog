Can LLMs "feel" music? After our class activity this past Monday, I was really curious to see how LLMs might describe and depict music. After all, it may be able to analyze the mood and tone of the pieces based on the musical notation of the piece to have a certain psychological effect on us. Perhaps it may be even more skilled at analyzing these pieces compared to an untrained ear considering it has a rich database that may contain analysis of musical pieces. I wanted to experiment with this idea myself, so I chose to listen to one of the most iconic classical pieces--Ride of the Valkyries by Richard Wagner.

### Part 1: My Reflection

To give a point of comparison, I listened to the piece in full-length first, reflecting on the following points to incorporate in my prompt:

    **1. How I feel when I listen to the music**
    **2. Points that stand out to me**
    **3. A story that resonates with me when listening to the music**
    
I then synthesized my reflection and asked gpt-4o to rewrite my reflection into an image prompt. I used the following prompt to execute this:

> **INPUT**: Please rewrite the following reflection into an image prompt for use with an LLM to generate an image: "Triumphant feeling with trumpets. Very dramatic. Army returns from a grueling war with battle scars and fatigue with a strong pride and feeling of victory. A crowd of people in a small town of primary colored german houses cheer for their return. It feels powerful, with dramatic lighting."
>
> *note: the output was revised to make it learn towards fantasy realism elements*
> 
> **OUTPUT**: A grand, painterly fantasy realism scene of a triumphant army returning home after a long and brutal war. The soldiers, weary yet proud, stride through a small, storybook German town filled with vividly painted, primary-colored houses. Their armor gleams with dents and scratches, capes flowing in the wind, faces marked with fatigue and honor. Townsfolk line the cobblestone streets, cheering and tossing flowers in celebration. Banners ripple overhead as sunlight pierces through dramatic clouds, bathing the scene in golden, radiant light. The atmosphere feels epic and emotional — a moment of glory and relief underscored by the imagined sound of triumphant trumpets.

### Part 2: LLM's Reflection

This next part involved prompting the LLMs (Gemini & GPT-5) to reflect on the song and generate an image. I gave the LLM the following prompt for this purpose:

> Please listen to this song: https://www.youtube.com/watch?v=GGU1P6lBW6Q. Based on the linked music, I want you to describe the following items:
>
> - how you feel when you listen to the music
> - points that stand out to you
> - a story that resonates with you when listening to the music
>  
> Given these three points, please write an image prompt that would consolidate these three points and create a visual depiction of the music.

#### **GPT-5**

I was thoroughly impressed by GPT-5's analysis of the piece. It particularly commented on the exhilarating nature of the piece and the excitement it invokes in its listeners. It also comments on how the music makes you feel both "tension and release," which then ultimately has the effect of making "you feel lifted, [carried, and transformed]."

Ultimately, GPT-5 gave me the following image prompt, which then generated the images (Gemini & GPT-5):

> A dynamic, heroic scene: a tall winged figure (angelic or Valkyrie‐like) soaring high above a dramatic, storm-scoured mountain valley. Massive brass-coloured wings span wide, catching the slanting light of a breaking dawn. Below, rugged peaks and dark clouds part to reveal golden rays and swirling mist. The figure is clad in flowing armor that reflects both classical and mythic motifs, the pose one of triumphant ascent: wings fully spread, one fist raised, head tilted up toward the light. The palette: deep storm blues and grays in the lower valley, transitioning to warm golds, amber and white light above. In the background, the remnants of battle: broken spears, ancient stone, scattered banners but these are partially obscured by lifting mist. The composition: sweeping diagonals that convey movement and lift, the wings and light forming a V-shape which echoes triumph. The mood: epic, liberating, powerful, mythic. ### Conclusion


This was such a fascinating activity and likely one of my most favorite blog posts I've written thus far! When comparing between the LLM's generations and refleciton of the music compared to mine, it feels as though their analyses are more "correct" as it likely draws from databases analysing this piece. Due to this, I do wonder how much of this output is a generalization of analysis of this musical piece completed by musicians prior, or the LLM's own independent generalization and analysis of the musical orchestration and patterns in this piece. The similar motifs and reflections generated by both Gemini and GPT-5 hints to me that it is likely the former, but it would be exciting to dive into this topic more deeply and learn more about LLM competency in the use of music theory.
