Remember the guessing genie craze almost a decade ago? That was Akinator, a digital genie who could seemingly read your mind and guess any character or celebrity you were thinking of by answering only a few yes-or-no questions. Akinator uses a combination of machine learning and algorithms based on large databases and decision trees to narrow down an answer. In this matter, the Akinator, despite its high accuracy and LLM-like features, is not an LLM but rather uses a highly refined binary search algorithm.

Reliving this guessing game craze, I wanted to compare an LLM versus Akinator to see which method would be more efficient in performing this task: guessing my character--Miffy.

---

### Comparing Akinator & Gemini

I played my first game with Akinator. Akinator was able to guess my character quite promptly within 25 questions. 

With Gemini 2.5 Flash, I used the following prompt to start the discussion:

> Let's play a game. You will ask me a series of yes-and-no questions to guess a character I'm thinking of. You can start.

Ultimately, in a round of 29 questions, Gemini was unable to deduce that my character was Miffy. Its final guess was Hazel from "The Countryside Adventures of Peter Rabbit." Gemini especially struggled with this character guessing, as it was hung up on whether the character was from a British children's show. One of its seemingly initial assumptions of the character was Ruby from "Max & Ruby," another popular kid's show featuring a rabbit protagonist with a younger brother--similar to Miffy as she has presented previously and will present in her future series. 

I thought that it was interesting that Gemini had summarized its findings based on my response to its yes-and-no questions, stating a variation of:

> Okay, not Ruby, and the character does not live in a natural dwelling like a burrow. This means they live in a more constructed home, even if their world is a wild/forest setting.
>
> Let's clarify that point:
>
> Does the character live in a house that resembles a human-style house (with walls, windows, furniture, etc.), rather than a simple hole or log?

for example. Notably, the fact that the LLM was unable to get past "not Ruby" tells me that the LLM almost was "thinking harder" because it couldn't find an answer that was not Ruby. This makes me wonder if the LLM's training data and its method of data retrieval and processing is not the most efficient for binary searches such as what this task involved. Perhaps if the LLM received instructions on how to search for this character, it would have been able to more efficiently pinpoint Miffy, however, it might be that the overabundance of data that the LLM has may also have a role in making this process less efficient. 

---

### Reflection

This was a fun game to experiment with the strengths and weaknesses of an LLM and to gain some insight into how these models process information, make inferences, and tap into their extensive knowledge bases. My main takeaway from this activity is that LLMs are not always the most efficient, and we should be thinking more deeply about which tasks would be most efficient with what form of algorithms or data processing.
